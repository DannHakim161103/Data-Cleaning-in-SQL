# BMI Calculator

```python
#BMI = weight in kilograms / (height in meters * height in meters)


# Under 18.5	Underweight	Minimal
# 18.5 - 24.9	Normal Weight	Minimal
# 25 - 29.9	Overweight	Increased
# 30 - 34.9	Obese	High
# 35 - 39.9	Severely Obese	Very High
# 40 and over	Morbidly Obese	Extremely High
```

```python
name = input("Enter your name: ")

weight = input("Enter your weight in kg: ")

height = input("Enter your height in meters: ")

BMI = float(weight) / (float(height) ** 2)
print(f"Your BMI is: {BMI}")

if BMI > 0:
    if BMI < 18.5:
        print(name + ", you are underweight"+ ", makan bnyak sikit la bro. ")
    elif 18.5 <= BMI < 25:
        print(name + ", you have a normal weight"+ ', mantap la bro.')
    elif 25 <= BMI < 30:
        print(name + ", you are overweight."+ ", jangan makan banyak la bro.")
    elif 30 <= BMI < 35:
        print(name + ", you are obese"+ ", Diet la bro.")
    elif 35 <= BMI < 40:
        print(name + ", you are severely obese"+ ", Diet la bro jangan makan bnyak ii.")
    else:
        print(name + ", you are morbidly obese "+ ", Diet la bro,jangan makan bnyak ii and exercise gak.")
else:
    print( " invalid input. Please enter valid weight and height values.")
```
